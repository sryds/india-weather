## Daily Weather from Top 10 Major Indian Cities, 2000-2024
### Detailed EDA at City Level

#### Import Libraries and set path

```python
import pandas as pd
import numpy as np
import os
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib
import matplotlib.ticker as ticker
from matplotlib.ticker import MultipleLocator
```

```python
path = r'C:\Users\artoe\Documents\DataAnalytics\projects\india weather\archive'
```

#### Import data as pickle

```python
df = pd.read_pickle(os.path.join(path, 'daily_weather_updated.pkl'))
```

### Create Subsets for Each City

```python
#find list of cities
df['city'].value_counts()
```

```python
#create a subset for each city
df_delhi =  df[df['city']=='Delhi']
df_mumbai =  df[df['city']=='Mumbai']
df_kolkata =  df[df['city']=='Kolkata']
df_chennai =  df[df['city']=='Chennai']
df_bangalore =  df[df['city']=='Bangalore']
df_hyderabad =  df[df['city']=='Hyderabad']
df_ahmedabad =  df[df['city']=='Ahmedabad']
df_pune =  df[df['city']=='Pune']
df_jaipur =  df[df['city']=='Jaipur']
df_lucknow =  df[df['city']=='Lucknow']
```

### EDA for Delhi

#### Daily Minimum and Maximum Temperatures (Monthly Average)

```python
melted_minmax_delhi = df_delhi.melt(
    id_vars='month_year',
    value_vars=["min_temp_monthly_avg", "max_temp_monthly_avg"],
    var_name="Temp_Type",
    value_name="Mean_Temp",
)
```

```python
melted_minmax_delhi["Temp_Type"] = melted_minmax_delhi["Temp_Type"].replace(
    {"min_temp_monthly_avg": "Daily Minimum", "max_temp_monthly_avg": "Daily Maximum"}
)
```

```python
fig, ax = plt.subplots(figsize=(20,5))
sns.lineplot(data=melted_minmax_delhi, x='month_year', y='Mean_Temp', hue='Temp_Type')
ax.xaxis.set_major_locator(ticker.MultipleLocator(12))
plt.title('Average Daily Minimum and Maximum Temperatures by Month, Delhi, 2000-2024')
plt.show()
```

#### Daily Mean and Perceived Daily Mean Temperatures (Monthly Average)

```python
melted_temps_delhi = df_delhi.melt(
    id_vars='month_year',
    value_vars=["mean_temp_monthly_avg", "apparent_mean_temp_monthly_avg"],
    var_name="Temp_Type",
    value_name="Mean_Temp",
)
```

```python
melted_temps_delhi["Temp_Type"] = melted_temps_delhi["Temp_Type"].replace(
    {"mean_temp_monthly_avg": "Daily Mean Temp (C)", "apparent_mean_temp_monthly_avg": "Perceived Daily Mean Temp (C)"}
)
```

```python
fig, ax = plt.subplots(figsize=(20,5))
sns.lineplot(data=melted_temps_delhi, x='month_year', y='Mean_Temp', hue='Temp_Type')
ax.xaxis.set_major_locator(ticker.MultipleLocator(12))
plt.title('Average Daily Mean and Perceived Daily Mean Temperatures by Month, Delhi, 2000-2024')
plt.show()
```

#### Daily Precipitation Levels (Monthly Average)

```python
melted_prec_delhi = df_delhi.melt(
    id_vars='month_year',
    value_vars=['precipitation_monthly_avg', 'rain_monthly_avg'],
    var_name='precipitation_type',
    value_name='precipitation_level',
)
```

```python
melted_prec_delhi['precipitation_type'] = melted_prec_delhi['precipitation_type'].replace(
    {'precipitation_monthly_avg': 'Average Daily Precipitation', 'rain_monthly_avg': 'Average Daily Rain'}
)
```

```python
fig, ax = plt.subplots(figsize=(20,5))
sns.lineplot(data=melted_prec_delhi, x='month_year', y='precipitation_level', hue='precipitation_type')
ax.xaxis.set_major_locator(ticker.MultipleLocator(12))
plt.title('Average Daily Precipitation vs Rain Levels by Month, Delhi, 2000-2024')
plt.show()
```

#### Daily Maximum Wind Speed and Wind Gusts (Monthly Average)

```python
melted_wind_delhi = df_delhi.melt(
    id_vars='month_year',
    value_vars=['max_wind_monthly_avg', 'max_windgusts_monthly_avg'],
    var_name='wind_gusts',
    value_name='wind_speed',
)
```

## General Observations

#### Cities that exhibit a wider seasonal variation of daily temperatures (e.g. Delhi, Jaipur) shows less difference between minimum and maximum temperatures, as well as between real and perceived temperatures. These cities also tend to experience higher levels or precipitation.

#### There appears to be no difference between daily precipitation and daily rain levels, indicating that other forms of precipitation, such as snow or sleet, did not occur at all in any of these cities. 

#### Chennai, Mumbai, Pune and Kolkata show increased maximum wind speeds and wind-gust speeds from 2017 onwards, while the remaining cities did not show any similar change in wind speeds. This increase in wind intensity may be linked to the coastal or near-coastal location of these 4 cities.

#### Different cities experienced higher than usual peaks in precipitation levels at different points during this time period. E.g. Delhi in summer 2016, Jaipur in summer 2013, Chennai in autumn 2021 and Ahmedabad in summer 2017.

#### All cities' precipitation levels peak at the same point each year, with the exception of Chennai. This is likely a reflection of the different monsoons experienced by these cities.
